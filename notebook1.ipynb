### Agregar una columna: importe de venta con descuento

```python
def calcular_importe(d):
    importe = d["cantidad"] * d["precio_unitario"]
    descuento = importe * d["descuento_pct"] / 100
    resultado = round(importe - descuento, 2)
    return resultado
```

```python
data["importe_neto"] = data.apply(calcular_importe, axis=1)
```

```python
data
```

## 2. Polars

Repetimos las operaciones sobre el mismo archivo CSV.

```python
import polars as pl
```

```python
datos_polars = pl.read_csv(f"{ruta}/ventas.csv")
```

```python
datos_polars
```

```python
datos_polars.schema
```

```python
datos_polars = datos_polars.with_columns(
    (pl.col("cantidad") * pl.col("precio_unitario") * (1 - pl.col("descuento_pct") / 100)).round(2).alias("importe_neto")
)
```

```python
datos_polars
```

## 3. Gráfico con Bokeh

Usamos Pandas para sumar el importe neto de las ventas entregadas por departamento.

```python
ventas = data[data["estado"] == "Entregado"]
```

```python
ventas
```

```python
resumen = ventas.groupby("departamento", as_index=False)["importe_neto"].sum()
```

```python
resumen = resumen.sort_values("importe_neto", ascending=False).head(8)
```

```python
resumen
```

```python
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
```

```python
output_notebook()
```

```python
grafico = figure(x_range=resumen["departamento"].tolist(), title="Ventas entregadas por departamento", width=800, height=350)
```

```python
grafico.vbar(x=resumen["departamento"].tolist(), top=resumen["importe_neto"].tolist(), width=0.7)
```

```python
grafico.xaxis.major_label_orientation = 0.8
grafico.yaxis.axis_label = "Importe neto (S/)"
```

```python
show(grafico)
```
